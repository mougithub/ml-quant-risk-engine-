```python
import pandas as pd
import matplotlib.pyplot as plt

from src.features.volatility import (
    calculate_rolling_volatility,
    calculate_ewma_volatility
)
```

```python
returns = pd.read_csv(
    "../data/processed/portfolio_returns.csv",
    index_col=0,
    parse_dates=True
)

portfolio_returns = returns.mean(axis=1)
```

```python
vol20 = calculate_rolling_volatility(
    portfolio_returns,
    window=20
)

ewma = calculate_ewma_volatility(
    portfolio_returns
)
```

```python
plt.figure(figsize=(12, 6))

vol20.plot(label="20D Rolling Vol")
ewma.plot(label="EWMA Vol")

plt.legend()
plt.title("Portfolio Volatility")
plt.show()
```
